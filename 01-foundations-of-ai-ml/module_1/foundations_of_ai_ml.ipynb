## **2. DATA**

### **1. Data Pipelines**

To unlock valuable insights and make accurate predictions, your models need high-quality, well-prepared data. This is where data pipelines and sources become essential. 

> A well-designed data pipeline keeps your data clean, organized, and analysis-ready.


### **2. Data Sources**

**Data sources = the origins where your project's data comes from before it enters any pipeline.**

**Formal definition:** Data sources are the origins from which you gather the data needed for AI/ML projects, ranging from structured databases to unstructured data (images, text) to real-time streams from sensors or APIs.

**Analogy:** Think of it like ingredients before cooking some come pre-chopped in labeled containers (structured), some are raw and messy (unstructured), some arrive fresh and continuously (real-time streams). Where you source them shapes how you prep them.

**The four types, for your notes:**

| **Type** | **Examples** | **Structure** |
|---|---|---|
| **Relational databases** | SQL Server, MySQL, PostgreSQL | Fixed schema, tables/rows/columns, queried via SQL |
| **NoSQL databases** | Azure Cosmos DB, MongoDB | Unstructured/semi-structured, flexible schema, key-value, document, or graph formats |
| **Cloud storage** | Azure Blob Storage, Amazon S3 | Scalable storage for large unstructured data (images, video, backups) |
| **Real-time streams** | IoT sensors, social media APIs | Continuous data, processed live for immediate insights |

**Why it matters:** the type of data source you're pulling from determines how you design the rest of the pipeline (ingestion method, storage choice, processing tools) it's the first decision point, not just a background detail.

### **3. What is a Data Pipeline?**

**Formal definition:** A data pipeline is a series of automated processes that move data from its source to a destination, transforming it along the way to prepare it for analysis or machine learning.

**Analogy:** A conveyor belt in a factory. Raw, unprocessed data enters the pipeline, moves through stages of transformation, and exits as clean, organized information ready for use.


### **4. Key Stages of a Data Pipeline**

- **Stage 1: Data Ingestion**
    **Role:** First stage, where raw data is collected from various sources. Can happen in real time (streaming) or in batches (bulk transfers), depending on the nature of the data and project requirements.

    **Tool:** Azure Data Factory. Connects to a wide range of sources (on prem databases, cloud storage, APIs), schedules and automates transfers to keep the pipeline up to date.

- **Stage 2: Data Processing / Transformation**
    **Role:** Raw ingested data is processed and transformed to make it suitable for analysis or model training: cleaning (removing duplicates, filling missing values), normalizing (scaling numerical values), transforming (aggregating data, joining tables).

    **Tool:** Azure Databricks. Collaborative environment built on Apache Spark, for large scale data processing and complex transformations.

- **Stage 3: Data Storage**
    **Role:** After processing, data is stored where it can be easily accessed by data analysts, data scientists, and AI/ML models. Storage choice depends on data type and downstream use.

    **Tools:**
    **Azure Data Lake Storage:** optimized for vast amounts of raw, unstructured data
    **Azure SQL Database:** ideal for structured data needing advanced querying
    **Azure Blob Storage:** versatile storage for unstructured data (text, images, video)

- **Stage 4: Data Access and Utilization**
    **Role:** Final stage. Processed, stored data is made available for analysis or as input into ML models. Data scientists and engineers use it to train models, generate reports, or drive decisions.

    **Tools:** Azure Machine Learning Service (model training), Power BI (business analytics)


### **5. Role of Data Pipelines in AI/ML Projects**

- Pipeline quality and reliability directly impact model performance
- A well structured pipeline means consistent, accurate, up to date data, resulting in reliable, effective models
- A weak pipeline means models trained on flawed data, leading to poor predictions and costly mistakes
- **Scalability:** a well designed pipeline handles growing data volume and variety, letting AI/ML systems scale. This is a competitive advantage in modern projects

## **4. Data Management**

### **4.1 SQL for Data Acquisition**

**SELECT**
Pulls specific columns from a table.
`SELECT name, email FROM customers WHERE country = 'US';`

**WHERE**
Filters rows by a condition.
`SELECT * FROM orders WHERE order_date > '2024-01-01';`

**JOIN**
Combines rows from two tables using a shared column.
- INNER JOIN: only matching rows in both tables
- LEFT JOIN: all rows from left table, matched or NULL from right
- RIGHT JOIN: all rows from right table, matched or NULL from left
- FULL JOIN: all rows from both, matched or not

**GROUP BY + Aggregation**
Groups rows and summarizes with COUNT, SUM, AVG, MIN, MAX.
`SELECT product_id, COUNT(*) FROM sales GROUP BY product_id;`

**ORDER BY**
Sorts results. ASC (default) or DESC.

**INSERT / UPDATE / DELETE**
INSERT adds a row. UPDATE modifies a row. DELETE removes a row.

**Subquery**
A query nested inside another query, used when one filter depends on the result of another lookup.

**Indexing**
Speeds up lookups on a column by keeping a sorted pointer to it. Costs a bit on write speed.

**Transactions (ACID)**
A group of SQL operations treated as one unit, either all succeed or none do. Keeps data consistent during multi-step operations like transfers.

- When you run `requests.get(url)` with no headers specified, the requests library still sends headers, it just fills in its own defaults. One of those defaults is:

    ```python
    User-Agent: python-requests/2.31.0
    ```
    > **Means:** "I am a Python script, version 2.31.0 of the requests library. Any server can read that header and immediately know you're not a browser

- An HTTP request isn't just "give me this page," it's a request with a bunch of metadata attached, sent as key-value pairs called headers. A few common ones:

  - **User-Agent**: who's asking (browser name, OS, version). This is the one that got you blocked.
  - **Accept**: what content types you're willing to receive (HTML, JSON, images, etc.)
  - **Referer**: what page you clicked from to get here
  - **Cookie**: session/login data, if any

- We're overriding just the `User-Agent` key so the server sees a string that matches what real Chrome sends, instead of `python-requests/2.31.0`. The server has no way to actually verify this, it just trusts the header, so this works by presenting a claim the server doesn't cross-check.

---

- `Mozilla/5.0`                → legacy tag, every browser still says this
- `AppleWebKit/537.36`         → rendering engine (used by Chrome & Safari)
- `(KHTML, like Gecko)`        → old compatibility claims (WebKit's ancestor + Firefox's engine)
- `Chrome/151.0.0.0`           → the real signal: this is Chrome, version 151
- `Safari/537.36`              → kept for WebKit compatibility checks

**What's happening, line by line:**

- `data = []` — an empty list to collect our cleaned-up rows.
- `for row in rows:` — go through each `<tr>` we found earlier, one at a time.
- `row.find_all('td')` — inside this one row, find all `<td>` (data cell) tags. Note: header rows use `<th>` not `<td>`, so this naturally skips the header row's cells.
- `col.text.strip()` — `.text` pulls just the visible text out of each cell (stripping away HTML tags), `.strip()` removes leading/trailing whitespace and newlines that HTML often leaves behind.
- `if cols:` — if a row had no `<td>` cells (like the header row, which only has `<th>`), `cols` would be an empty list, this skips adding it.
- `data.append(cols)` — add this row's cleaned data to our master list.

### **2. Outliers**
**Analogy:** One student scoring 5 when everyone else scores 60-80.

- **Detect:** Z-score (>3 std devs away) or IQR (outside Q1-1.5×IQR to Q3+1.5×IQR)
- **Handle:** Remove, or cap/transform (e.g. log)

```python
z_scores = np.abs(stats.zscore(df['col']))
```

````python
# All in function 
def load_dataset_titanic():
    return pd.read_csv('../module_1/csv/Titanic-Dataset.csv')

def handle_missing_values(df):
    clean_df = df
    msno.matrix(df)
    msno.heatmap(df)
    clean_df = clean_df.drop(columns=['Cabin', 'Name', 'Ticket', 'PassengerId'])
    clean_df['Age'] = clean_df['Age'].fillna(clean_df['Age'].median())
    clean_df['Embarked'] = clean_df['Embarked'].fillna(clean_df['Embarked'].mode()[0])
    return clean_df

def remove_outliers(df):
    numeric_cols = df.select_dtypes(include=[np.number])
    zscores = np.abs(stats.zscore(numeric_cols))
    df_no_outliers = numeric_cols[(zscores < 3).all(axis=1)]
    df_cleaned = df.loc[df_no_outliers.index]
    return df_cleaned

def scale_data(df):
    numeric_cols_data = df.select_dtypes(include=[np.number])
    scaler = StandardScaler()
    scale_data_df = pd.DataFrame(scaler.fit_transform(numeric_cols_data),
                                 columns=numeric_cols_data.columns,
                                 index=numeric_cols_data.index)
    df_final = df.copy()
    df_final[scale_data_df.columns] = scale_data_df
    return df_final

def encode_categorical_data(df, categorical_columns):
    return pd.get_dummies(df, columns=categorical_columns)

# Example Usage:
df = load_dataset_titanic()
df_clean = handle_missing_values(df)
df_no_outliers = remove_outliers(df_clean)
df_scale_data = scale_data(df_no_outliers)
df_encode = encode_categorical_data(df_scale_data, 
                                    categorical_columns=['Sex', 'Embarked']) 
df_final = df_encode.copy()
df_final.head()  
````
