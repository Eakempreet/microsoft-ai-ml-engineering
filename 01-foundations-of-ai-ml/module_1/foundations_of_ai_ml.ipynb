# **Foundations of AI and Machine Learning** 🤖

### **Azure ML Pipeline:** 

**```Store → Prep → Build → Serve → Watch```**

```python
1. RAW DATA LANDS
   Azure Data Lake Storage     → messy files (images, logs)
   Azure SQL Database          → clean structured tables (your metadata CSV equivalent)

2. DATA GETS MOVED & PREPPED
   Azure Data Factory          → schedules/moves data between systems (the plumber)
   Azure Databricks            → heavy cleaning/transformation at scale (the muscle)

3. MODEL GETS BUILT
   TensorFlow                  → the actual framework you write the model in
   Azure ML Service            → the control room managing the whole ML lifecycle
   Azure ML SDK                → your Python remote control into that control room

4. MODEL GETS DEPLOYED
   Azure App Services          → simple hosting (HF Spaces equivalent)
   Azure Kubernetes Service    → hosting at massive scale, many containers

5. DEPLOYED MODEL GETS WATCHED
   Azure Monitor               → the watchdog, alerts you when it breaks
```
