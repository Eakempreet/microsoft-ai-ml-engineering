## **Azure ML Pipeline: Store → Prep → Build → Serve → Watch**
#### 1. STORE (where data lives)

**Azure Data Lake Storage**
> Think: a warehouse dumping ground for messy stuff like images, logs, anything unstructured.

Formal: Scalable cloud storage for large volumes of raw, unstructured, or semi-structured data.

**Azure SQL Database**
> Think: a neat filing cabinet, clean rows and columns, like your metadata CSV.

Formal: Fully managed relational database service for structured, queryable data.

#### 2. PREP (get data ready)

**Azure Data Factory**
> Think: the plumber, moves data from A to B on a schedule, no processing.

Formal: Cloud based ETL(Extract, Transform, Load)/orchestration service for moving and scheduling data between systems.

**Azure Databricks**
> Think: the muscle, heavy cleaning/transforming of huge datasets.

Formal: Apache Spark based analytics platform for large scale data processing and transformation.

#### 3. BUILD (create the model)

**TensorFlow**
> Think: the toolbox you build the neural network with.

Formal: Open source deep learning framework used to build and train neural networks.

**Azure Machine Learning Service**
> Think: the control room, manages the whole ML lifecycle in one place.

Formal: Managed cloud platform for the end to end ML lifecycle (training, registry, deployment, monitoring).

**Azure Machine Learning SDK**
> Think: your Python remote control into that control room.

Formal: Python library for interacting with Azure ML Service programmatically.

#### 4. SERVE (deploy as a web service)

**Azure App Services**
> Think: HF Spaces equivalent, simple hands off web app hosting.

Formal: Managed PaaS(Platform as a Service) for hosting web applications/APIs, including ML models, without managing infrastructure.

**Azure Kubernetes Service (AKS)**
> Think: hosting at massive scale, many containers, load balanced, auto healing.

Formal: Managed container orchestration (the automated coordination, scheduling, and management of complex data pipelines and tasks across multiple systems) service for deploying and scaling containerized apps at scale.

#### 5. WATCH (keep it healthy)

**Azure Monitor**
> Think: the watchdog, flags you the moment something breaks.

Formal: Service for collecting, analyzing, and acting on telemetry (logs, metrics, alerts).

## **DATA**

### **1. Data Pipelines**

To unlock valuable insights and make accurate predictions, your models need high-quality, well-prepared data. This is where data pipelines and sources become essential. 

> A well-designed data pipeline keeps your data clean, organized, and analysis-ready.


### **2. Data Sources**

**Data sources = the origins where your project's data comes from before it enters any pipeline.**

**Formal definition:** Data sources are the origins from which you gather the data needed for AI/ML projects, ranging from structured databases to unstructured data (images, text) to real-time streams from sensors or APIs.

**Analogy:** Think of it like ingredients before cooking some come pre-chopped in labeled containers (structured), some are raw and messy (unstructured), some arrive fresh and continuously (real-time streams). Where you source them shapes how you prep them.

**The four types, for your notes:**

| **Type** | **Examples** | **Structure** |
|---|---|---|
| **Relational databases** | SQL Server, MySQL, PostgreSQL | Fixed schema, tables/rows/columns, queried via SQL |
| **NoSQL databases** | Azure Cosmos DB, MongoDB | Unstructured/semi-structured, flexible schema, key-value, document, or graph formats |
| **Cloud storage** | Azure Blob Storage, Amazon S3 | Scalable storage for large unstructured data (images, video, backups) |
| **Real-time streams** | IoT sensors, social media APIs | Continuous data, processed live for immediate insights |

**Why it matters:** the type of data source you're pulling from determines how you design the rest of the pipeline (ingestion method, storage choice, processing tools) it's the first decision point, not just a background detail.

### **3. What is a Data Pipeline?**

**Formal definition:** A data pipeline is a series of automated processes that move data from its source to a destination, transforming it along the way to prepare it for analysis or machine learning.

**Analogy:** A conveyor belt in a factory. Raw, unprocessed data enters the pipeline, moves through stages of transformation, and exits as clean, organized information ready for use.


### **4. Key Stages of a Data Pipeline**

- **Stage 1: Data Ingestion**
    **Role:** First stage, where raw data is collected from various sources. Can happen in real time (streaming) or in batches (bulk transfers), depending on the nature of the data and project requirements.

    **Tool:** Azure Data Factory. Connects to a wide range of sources (on prem databases, cloud storage, APIs), schedules and automates transfers to keep the pipeline up to date.

- **Stage 2: Data Processing / Transformation**
    **Role:** Raw ingested data is processed and transformed to make it suitable for analysis or model training: cleaning (removing duplicates, filling missing values), normalizing (scaling numerical values), transforming (aggregating data, joining tables).

    **Tool:** Azure Databricks. Collaborative environment built on Apache Spark, for large scale data processing and complex transformations.

- **Stage 3: Data Storage**
    **Role:** After processing, data is stored where it can be easily accessed by data analysts, data scientists, and AI/ML models. Storage choice depends on data type and downstream use.

    **Tools:**
    **Azure Data Lake Storage:** optimized for vast amounts of raw, unstructured data
    **Azure SQL Database:** ideal for structured data needing advanced querying
    **Azure Blob Storage:** versatile storage for unstructured data (text, images, video)

- **Stage 4: Data Access and Utilization**
    **Role:** Final stage. Processed, stored data is made available for analysis or as input into ML models. Data scientists and engineers use it to train models, generate reports, or drive decisions.

    **Tools:** Azure Machine Learning Service (model training), Power BI (business analytics)


### **5. Role of Data Pipelines in AI/ML Projects**

- Pipeline quality and reliability directly impact model performance
- A well structured pipeline means consistent, accurate, up to date data, resulting in reliable, effective models
- A weak pipeline means models trained on flawed data, leading to poor predictions and costly mistakes
- **Scalability:** a well designed pipeline handles growing data volume and variety, letting AI/ML systems scale. This is a competitive advantage in modern projects


### **6. Types of ML Using Data Pipelines**

| **Type** | **Data Used** | **Goal** |
|---|---|---|
| **Supervised learning** | Labeled data (input output pairs) | Learn from known outcomes: classification, regression |
| **Unsupervised learning** | Unlabeled data | Find patterns/groupings (clustering, association), exploratory, hidden structure |
| **Reinforcement learning** | No fixed dataset, agent + environment | Learn via trial and error, maximizing rewards: game playing, robotics |

**Note:** which type you use depends on labeled data availability and task complexity.

## **ML Frameworks Comparison**

| Framework | Type | Key Features | Best For |
|---|---|---|---|
| **TensorFlow** | Deep learning | Eager + graph execution, scales on CPU/GPU/TPU, TensorFlow Serving | Image recognition, NLP, reinforcement learning |
| **PyTorch** | Deep learning | Dynamic computation graphs, Pythonic, TorchScript for deployment | Research, prototyping, computer vision, NLP (Transformers) |
| **Scikit-learn** | Traditional ML | Simple API, wide algorithm range, built in evaluation tools | Predictive modeling, data preprocessing, model evaluation |
| **Azure ML SDK** | Cloud/enterprise ML | AutoML, Azure integration, experiment tracking, scalable deployment | Enterprise AI, AutoML, model monitoring |
