# **🗃️ Which framework is best for the model?**

## **1. Overview of ML frameworks and libraries**


## **1.1 Introduction to popular ML frameworks**

### **What is a framework?**

A framework is a toolkit of prebuilt code for **building, training and deploying** models, so you do not write the algorithms from scratch.

**Analogy:** A framework is a fully equipped kitchen. You still decide the recipe and do the cooking, but you do not forge the pans.

```
Without a framework:  write gradient descent, layers, data loaders by hand
With a framework:     model.fit(data) and the heavy lifting is done for you
```

---

### **The five to know**

| Framework | Built by | Best for | Main strength |
|---|---|---|---|
| **TensorFlow** | Google Brain | Production deep learning | Scales across CPUs, GPUs, TPUs and mobile |
| **PyTorch** | Facebook AI Research | Research and prototyping | Dynamic computation graph, very Pythonic |
| **Scikit learn** | Open source (built on NumPy, SciPy) | Classical ML | Simple, consistent API |
| **Keras** | Runs on top of TensorFlow | Fast deep learning builds | High level, modular, pretrained models |
| **Spark MLlib** | Apache Spark | Big data | Distributed, streaming, real time |

---

### **1. TensorFlow**

**In one line:** The industrial strength choice for taking deep learning from experiment to real product.

**Key ideas explained:**
* **CPU, GPU, TPU:** three kinds of processors. CPU is general purpose, GPU does many calculations in parallel (ideal for neural networks), TPU is Google's chip built only for neural networks. TensorFlow runs on all three, and on phones too.
* **Keras inside:** the easy interface for building models is bundled in.
* **TensorBoard:** a dashboard that plots loss and accuracy while a model trains, so you can spot problems early.

**Use cases:** image and speech recognition, NLP, time series, large scale neural network training

**Pick it when:** the model will ship to production or needs to scale.

---

### **2. PyTorch**

**In one line:** The researcher's favourite, built for experimenting quickly.

**Key ideas explained:**
* **Computation graph:** the map of calculations a network performs.
* **Dynamic graph:** PyTorch builds this map as the code runs, so you can change the model on the fly and debug it like normal Python. A static graph is fixed before running, which is harder to tweak.
* **Pythonic:** it reads like ordinary Python, so it feels natural to learn.

**Use cases:** research and prototyping, computer vision, reinforcement learning, custom neural networks

**Pick it when:** you are trying new ideas and need flexibility.

---

### **3. Scikit learn**

**In one line:** The standard toolbox for classical ML on tabular data.

**Key ideas explained:**
* **Classical ML:** algorithms like Random Forest, linear regression and clustering, as opposed to deep neural networks.
* **Consistent API:** almost every model uses the same pattern: `fit()`, then `predict()`. Learn one, you know them all.
* **Works with pandas and Matplotlib:** load data with pandas, model with scikit learn, plot with Matplotlib.
* **Dimensionality reduction:** shrinking many features into fewer while keeping the important information.

**Use cases:** predictive modeling, preprocessing, model evaluation, feature selection

**Pick it when:** the data is a table and fits in one machine's memory. 

---

### **4. Keras**

**In one line:** The easy front door to deep learning.

**Key ideas explained:**
* **High level API:** you describe the network in a few lines (add a layer, pick an optimizer) and it handles the complexity underneath.
* **Modular:** layers, optimizers, loss functions and metrics are swappable blocks.
* **Pretrained models:** networks already trained on huge datasets that you can reuse.
* **Transfer learning:** taking a pretrained model and adapting it to your task with little data. Your ATAS aircraft classifier used exactly this idea.

**Important:** Keras is **not a separate engine**. It is an interface that sits on top of a backend (TensorFlow is the main one).

**Use cases:** rapid prototyping, transfer learning, image classification, text and time series

**Pick it when:** you want to build deep learning models quickly.

---

### **5. Spark MLlib**

**In one line:** The ML library for data too big for one computer.

**Key ideas explained:**
* **Apache Spark:** an engine that splits huge data across many machines (a **cluster**) and processes the pieces in parallel. This is called **distributed processing**.
* **Hadoop and Hive:** big data storage and query tools many companies already use. MLlib plugs into them.
* **Streaming:** handling data as it arrives, which allows real time ML.

**Use cases:** big data analytics, real time ML, large scale data processing, distributed modeling

**Pick it when:** the data is so large that one machine cannot hold or process it.

---

### **Quick chooser**

```
Table data, small to medium size      → Scikit learn
Deep learning, going to production    → TensorFlow (with Keras)
Deep learning, research and ideas     → PyTorch
Data too big for one machine          → Spark MLlib
```

---

### **Remember**
* TensorFlow leans to **production scale**, PyTorch to **research flexibility**
* Keras is the **easy interface**, not a rival engine
* Scikit learn is for **classical ML**, Spark MLlib is for **big data**
* Choose by the **problem and its scale**, not by popularity
* Frameworks can be **combined**: Spark prepares huge data, TensorFlow trains the model

## **1.2 Tokens, self attention and transformers**

---

### **Tokens**

**What they are:** The small pieces of text a model actually reads. A token can be a whole word, part of a word, or a single character.

**Analogy:** Lego bricks. The model never sees a sentence, only bricks it has learned to snap together.

```
Text:    "Unbelievable results"
Tokens:  ["Un", "believ", "able", " results"]
IDs:     [3118, 28903, 540, 2482]
```

* Models split rare words into pieces, so they can handle words they have never seen
* Rough English rule: 1 token is about 4 characters, or 0.75 of a word
* **Context window** = the maximum number of tokens the model can read at once

**Full journey of your text:**

```
Text → Tokens → IDs → Embeddings (vectors of numbers) → Transformer → Next token probabilities
```

---

### **Self attention**

**What it is:** A mechanism where every token looks at all other tokens in the input and decides how much each one matters for understanding it.

**Example:**

```
"The animal did not cross the street because it was too tired"

What does "it" mean?
Self attention gives "it" a high weight toward "animal"
and a low weight toward "street"
```

**How it works, in three roles per token:**

| Role | Meaning | Library analogy |
|---|---|---|
| **Query** | What am I looking for? | Your search question |
| **Key** | What do I contain? | The label on each book |
| **Value** | The actual information | The content inside the book |

```
1. Compare each Query with every Key   → relevance scores
2. Softmax turns scores into weights   → they sum to 1
3. Take a weighted mix of the Values   → new meaning aware vector for the token
```

**Why it matters:** The model links words that are far apart in a sentence, which older models struggled with.

---

### **Transformer**

**What it is:** The neural network architecture, introduced in the 2017 paper "Attention Is All You Need", that is built around self attention. Nearly every modern LLM uses it.

**Analogy:** A team meeting where every person (token) hears everyone else at the same time, updates their understanding, then the meeting repeats for many rounds.

**One transformer block:**

```
Input vectors
    ↓
Self attention      (tokens share information)
    ↓
Feed forward layer  (each token processes what it gathered)
    ↓
Output vectors      (richer, context aware)

Stack this block dozens of times = a transformer
```

**Why it won:**
* Processes all tokens **in parallel**, so training is fast on GPUs (older RNNs read one word at a time)
* Captures long distance relationships
* Scales well with more data and parameters

**Order of words:** Attention alone ignores word order, so **positional encoding** is added to tell the model where each token sits.

**Three flavours (links back to your table):**

| Type | Reads | Examples |
|---|---|---|
| Encoder only | Whole text both directions | BERT, RoBERTa |
| Decoder only | Left to right, predicts next token | GPT, Codex |
| Encoder and decoder | Reads input, then writes output | T5 |

---

### **Remember**
* Tokens are the **units**, self attention is the **mechanism**, the transformer is the **architecture**
* Self attention answers: "which other tokens matter for this token?"
* Query, Key, Value is a search system inside the network
* Parallel processing is why transformers replaced RNNs

---

## **1.3 Overview of pretrained LLMs**

**What they are:** Deep learning models (mostly transformers) trained on huge amounts of text, so they already understand language before you touch them.

**Analogy:** A graduate who has read the whole library. You do not teach them language, you only train them for your specific job.

---

### **Key traits**
* **Scale:** billions of parameters
* **Versatility:** one model adapts to many tasks
* **Context:** understands meaning from surrounding words

---

### **How they work**

```
Stage 1: PRETRAINING   huge text corpus, learn to predict the next word
Stage 2: FINE TUNING   small task dataset, adapt to one job
Core engine: transformer with self attention (weighs which words matter)
```

---

### **Popular models**

| Model | By | Type | Best for |
|---|---|---|---|
| **GPT** | OpenAI | Generative | Text generation, chatbots, summarizing |
| **BERT** | Google | Understanding (masked words) | Q&A, sentiment analysis |
| **T5** | Google | Everything as text in, text out | Translation, summarizing, Q&A |
| **RoBERTa** | Facebook AI | Improved BERT | Classification, entity recognition |
| **Codex** | OpenAI | GPT descendant for code | Code generation and completion |

**Quick memory hook:**
* GPT writes, BERT reads
* T5 turns every task into text to text
* RoBERTa is BERT trained harder (more data, no next sentence prediction)

---

### **Benefits vs challenges**

| Benefits | Challenges |
|---|---|
| Less training time and resources | Heavy compute needed |
| Adapts to many tasks | Inherits bias from data |
| Strong benchmark performance | Hard to interpret |

---

### **Remember**
* Pretrain first, fine tune second
* Transformers and self attention power almost all of them
* Pretrained means you build on top, not from scratch

## **1.3 CNNs vs RNNs vs LLMs**

---

### **First, a correction to the framing**

CNN and RNN are **architectures** (how a network is wired). An LLM is a **type of model** (a big language model), and almost every LLM is built on the **transformer** architecture.

```
Architectures:   CNN      RNN      Transformer
                                       ↓
Model type:                           LLM
```

So the real comparison is **CNN vs RNN vs Transformer**, with LLMs as the famous product of the transformer.

---

### **CNN (Convolutional Neural Network)**

**Built for:** Images and grid shaped data

**Analogy:** A detective with a small magnifying glass, sliding it across a photo to spot edges, then shapes, then objects.

```
Pixels → Edges → Textures → Parts (wing, engine) → Whole object (F16)
```

* Uses small **filters** that slide over the image and detect patterns
* The same filter is reused everywhere, so it needs fewer parameters
* Learns patterns in layers, simple first, complex later
* **Weakness:** only sees a local area at a time, and has no sense of sequence

**You already used this:** your ATAS aircraft classifier is a CNN family model (EfficientNetV2).

---

### **RNN (Recurrent Neural Network)**

**Built for:** Sequences (text, speech, time series)

**Analogy:** Reading a book one word at a time while keeping a small notepad of what you have read so far.

```
Word 1 → [memory] → Word 2 → [memory] → Word 3 → ...
          ↑ passes a hidden state forward each step
```

* Processes tokens **one after another**
* Carries a **hidden state** as its memory
* Variants: LSTM and GRU, which remember longer
* **Weaknesses:**
  * Slow, because it cannot run in parallel
  * Forgets early words in long sequences
  * Hard to train on long text (vanishing gradients)

---

### **LLM (built on the Transformer)**

**Built for:** Language, now also code, images and more

**Analogy:** A team meeting where everyone hears everyone at once, instead of passing a note down a line.

* Uses **self attention**, so every token looks at every other token directly
* Processes all tokens **in parallel**, so it trains fast on GPUs
* Handles long range relationships well
* Scaled to billions of parameters and pretrained on huge text
* **Weaknesses:** heavy compute, hard to interpret, attention cost grows with input length

---

### **Side by side**

| | **CNN** | **RNN** | **LLM (Transformer)** |
|---|---|---|---|
| **Data type** | Images, grids | Sequences | Sequences, mainly text |
| **Core idea** | Sliding filters | Hidden state memory | Self attention |
| **Reads data** | Local patches | One step at a time | All tokens at once |
| **Long range context** | Weak | Weak to medium | Strong |
| **Parallel training** | Yes | No | Yes |
| **Typical use** | Image classification, detection | Older speech and text models | Chatbots, translation, code |
| **Size today** | Small to medium | Small to medium | Very large |

---

### **Quick chooser**

```
Is the input an image or video frame?     → CNN
Small sequence task, limited compute?     → RNN / LSTM (still works)
Language, long context, generation?       → Transformer / LLM
Image + language together?                → Transformer based multimodal model
```

**Note:** The lines blur. Vision Transformers (ViT) now do image work, and some CNN ideas appear inside newer models. Treat the table as the classic picture, not a strict wall.

---

### **Remember**
* CNN **looks locally**, RNN **remembers sequentially**, Transformer **attends globally**
* RNNs lost mainly because they were slow and forgetful
* LLM is a model type, the transformer is its engine

---

## **1.4 Practice: selecting a framework for a recommender**

### **The problem**
A retailer wants to recommend products to each customer using purchases, browsing behaviour and demographics.

---

### **Step 1: Understand the data**

| Data | Format | Meaning |
|---|---|---|
| Transactions, product info, ratings, demographics | **Structured** | Neat rows and columns, like a table |
| Browsing behaviour | **Semi structured** | Event logs in **JSON**: a session ID plus a list of actions (click, time on page, bought or left) |

* **Structured:** fits a table directly
* **Semi structured:** has some organisation (labelled fields) but no fixed table shape, so it must be converted into features before a model can use it
* **Scale:** millions of customers and events, so the data is **large and growing**, which makes scalability a hard requirement

**Output needed:** for one customer, a **ranked list of products** they are most likely to want.

---

### **Step 2: Choose the model type**

Two sources of knowledge can link a TV to a sound system:

| Type | Knowledge comes from | Example | Weakness |
|---|---|---|---|
| **Content based** | Product attributes (category, brand, features, price) | "This soundbar suits TVs" | Similar specs do not mean people buy them together |
| **Collaborative** | Other people's behaviour | "People who bought this TV also bought this soundbar" | **Cold start** |

**Cold start problem:** a brand new product or customer has no history, so collaborative filtering has nothing to learn from.

**The fix is a hybrid model:** combine both and each covers the other's weak spot.

```
Content model:        product attributes + customer demographics
Collaborative model:  purchases, ratings, browsing behaviour
Final output:         both scores combined into one ranked list
```

New soundbar with zero sales? The content side still scores it.

**Model families you may meet:**
* **Matrix factorization:** the classic maths version of collaborative filtering
* **Neural collaborative filtering (NCF):** the deep learning upgrade, learns richer patterns

---

### **Step 3: Match requirements to frameworks**

```
Requirement                       Points to
Large, growing data               Spark MLlib   (distributed processing across machines)
Hybrid neural model               TensorFlow    (deep learning, TF Recommenders)
Scale in production               Both can
```

**Primary choice: TensorFlow**
* Supports NCF and hybrid neural models
* Scales across CPUs, GPUs, TPUs and cloud
* **TensorFlow Recommenders (TFRS)** is a library built for this exact problem
* Large community and strong documentation

**Optional support: Spark MLlib**
* Prepares the huge logs in a distributed way, such as converting browsing JSON into structured features
* Then TensorFlow trains the model

---

### **Why not the others**

| Framework | Reason |
|---|---|
| **Scikit learn** | Built for small to medium data on one machine, no distributed computing |
| **Keras** | Not a separate engine, it is the easy interface on top of TensorFlow |
| **PyTorch** | A valid alternative, but it has fewer official ready made recommender tools and TensorFlow is the safer production fit |
| **Spark MLlib alone** | Strong at collaborative filtering and big data, but not built for deep learning |

---

### **Justification template**
> I select TensorFlow because the core of the task is a hybrid recommendation model combining product attributes with customer interaction data, which needs deep learning. It scales across CPUs, GPUs and TPUs, has strong community support, and TensorFlow Recommenders gives ready tools for this problem. Spark MLlib can support it by processing the very large transaction and browsing data.

---

### **Remember**
* Split models by **type of knowledge**, not by data format
* Choose a framework by **requirements first**, popularity last
* A justification needs the **why**, linked to a specific requirement
* Being able to do a task is not the same as being the best fit for it

## **2. Strengths and weaknesses of ML frameworks**

### 2.1 Various Frameworks

Besides TensorFlow, PyTorch and Keras (TensorFlow's default high level API), there are other frameworks, each with its own strengths and weaknesses: scikit-learn (classical ML), Apache MXNet and Caffe (deep learning).

#### Apache MXNet

**What it is:** An open source deep learning framework built for efficiency, flexibility and scalability.

**Core idea: hybrid programming**
* **Imperative** (dynamic graph): code runs line by line, easy to debug and change.
* **Symbolic** (static graph): the whole computation is declared first and optimized, so it runs faster.
* MXNet mixes both: experiment imperatively, then convert to symbolic for speed.

**Strengths**
* Hybrid model: flexibility of dynamic graphs plus speed of static graphs.
* Scalability: built for distributed training (one job split across many GPUs or machines).
* Performance: efficient memory use and speed on large datasets.
* Language support: Python, Scala, Julia, R.

**Weaknesses**
* Smaller community, so fewer tutorials and third party libraries.
* Steeper learning curve because of the hybrid model.
* Less popular, so weaker industry adoption.

**Best for:** Large scale distributed deep learning.

#### Caffe

**What it is:** A deep learning framework from the Berkeley Vision and Learning Center (BVLC), known for fast training of CNNs (Convolutional Neural Networks, used for images).

**Core idea: configuration driven**
* The network is described in a config file instead of regular code.
* This keeps it fast and lean, but less intuitive for code first users.

**Strengths**
* Speed: one of the fastest for training CNNs.
* Model Zoo: a collection of pretrained models you can reuse or fine tune.
* Efficiency: low memory overhead, good for limited compute.

**Weaknesses**
* Limited flexibility: specialized for CNNs.
* Slowed development: not actively maintained like TensorFlow or PyTorch.
* Less user friendly: config approach has a steeper learning curve for code first users.

**Best for:** Fast image model training on limited resources (mostly a legacy choice today).

#### Quick comparison

| Framework | Main strength | Main weakness |
|---|---|---|
| MXNet | Scalability plus hybrid graphs | Small community |
| Caffe | CNN speed, Model Zoo | Inflexible, slowed development |

### **2.1 How to choose the framwork?**

`Q1. What kind of problem?`
    Tables of numbers, classical ML  → Scikit-learn
    Images, text, deep learning      → go to Q2

`Q2. What stage are you at?`
    Learning or quick prototype      → Keras
    Research, custom new ideas       → PyTorch
    Shipping to real users           → TensorFlow

`Q3. Special scale or hardware?`
    Huge, many machines              → MXNet
    Fast CNN on limited compute      → Caffe


**For example:**

- Predict house price from a spreadsheet   → Scikit-learn
    (rows and columns, no deep learning needed)

- Cat vs dog image classifier, first try   → Keras
    (few lines, beginner friendly)

- Test a new neural network idea for a     → PyTorch
    research paper                              (dynamic graph, change on the fly)

- Image model inside a phone app           → TensorFlow
    (TensorFlow Lite for mobile)

- Train one giant model across 100 GPUs    → MXNet
    (built for distributed training)

- Fast image recognition, tight hardware   → Caffe
    (speed and low memory)

### **2.2 Real world case studies of ML frameworks**

#### **TensorFlow at Google Health: diabetic retinopathy**
* **Problem:** detect an eye disease from retinal images, especially where few eye specialists exist.
* **Approach:** a CNN trained on millions of expert labeled images, scaled across multiple GPUs.
* **Outcome:** accuracy comparable to board certified ophthalmologists, used in screening programs.
* **Why TensorFlow:** scalability plus strong deep learning support.

#### **PyTorch at Tesla: autonomous driving**
* **Problem:** improve object detection and decisions for Autopilot.
* **Approach:** experimented with CNN and RNN architectures using the dynamic graph, trained on large labeled driving videos with GPU acceleration.
* **Outcome:** fewer false positives and negatives in object detection.
* **Why PyTorch:** flexibility, rapid prototyping, strong GPU support.

#### **Scikit-learn at JPMorgan Chase: credit risk**
* **Problem:** predict which loan applicants may default, with decisions that can be explained.
* **Approach:** preprocessing with StandardScaler (scaling numbers) and OneHotEncoder (turning categories into numbers), logistic regression as the final model, tuned with GridSearchCV (tests many hyperparameter settings with cross validation).
* **Outcome:** reported 15 percent drop in loan default rates.
* **Why Scikit-learn:** simple, interpretable classical ML on tabular data.

#### **Pattern to remember**
| Framework | Data type | Deciding strength |
|---|---|---|
| TensorFlow | Images, large scale | Scalability |
| PyTorch | Video, fast changing models | Flexibility |
| Scikit-learn | Tables | Interpretability |
